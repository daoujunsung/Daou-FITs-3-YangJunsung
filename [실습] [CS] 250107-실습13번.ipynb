```python
# 실습과제13 파이썬을 이용한 배열 실습

# 배열의 요소 추가/삽입
arr = [1,2,3]
arr.append(4) # [1,2,3,4]
print(arr)
arr.insert(1,10) # [1,10,2,3,4]
print(arr)
arr_ones = [1]*5 # [1,1,1,1,1]
print(arr_ones)
```

```python
# 배열 요소 수정
arr = [1,2,3,4,5]
arr[2] = 99 # [1,2,99,4,5]
print(arr)
```

```python
# 배열 요소 삭제
arr = [10,20,30,40,50]
arr.remove(30) # [10,20,40,50]
print(arr)
arr.pop() # [10,20,40] (마지막 요소 삭제)
print(arr)
arr.pop(0) # [20,40] (첫번째 요소 삭제)
print(arr)
```

```python
# 배열 합치기
arr1 = [1,2,3]
arr2 = [4,5,6]
arr3 = arr1 + arr2 # [1,2,3,4,5,6]
print(arr3)
arr1.extend(arr2) # [1,2,3,4,5,6]
print(arr1)
```
